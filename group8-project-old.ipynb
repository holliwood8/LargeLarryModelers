```python
df.rename(columns={
    'HOMA-IR (insulin resistance)_None': 'HOMA_IR',
    'High Sensitivity CRP (mg/L)_mg/L': 'hsCRP',
    'Lp-PLA2_None': 'Lp_PLA2',
    'Fasting Glucose_mg/dL': 'fasting_glucose'
}, inplace=True)

data = df[['Date', 'hsCRP', 'Lp_PLA2', 'fasting_glucose']]
data
```

```python
data_long = pd.read_csv("Data/LLM_Biomarkers_Long.csv")
data_long.head()

data_long['Panel'].value_counts()

# filter for Inflammation Panel and Glucose/Insulin Panel 
data_filtered = data_long[data_long['Panel'].isin(['Inflammation Panel', 'Glucose/Insulin Panel'])]
data_filtered.head()

value_num = (data_filtered['Value']
             .astype(str)
             .str.replace(',', '.', regex=False)        # convert 0,5 → 0.5
             .str.extract(r'([-+]?\d*\.?\d+)')[0])      # pull out the numeric token
data_filtered['Value_numeric'] = pd.to_numeric(value_num, errors='coerce')

data_pivot = (
    data_filtered
    .pivot_table(index='Date',
                 columns='Biomarker',
                 values='Value_numeric',
                 aggfunc='mean')
    .reset_index()
)
# pivot the data such that each row is a date and each column is a biomarker
# data_pivot = data_filtered.pivot_table(index='Date', columns='Biomarker', values='Value')
# data_pivot.reset_index(inplace=True)
# data_pivot.head()

data_pivot.head()
```
